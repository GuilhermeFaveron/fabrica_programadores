## Exercício 1 — Dissídio anual de uma equipe

O departamento de Recursos Humanos possui os seguintes salários:

```python
salarios = np.array([1850, 2300, 2750, 3200, 4100, 5250, 6800, 8400])
```

A convenção coletiva determinou um **dissídio de 5,7%**.

Utilizando NumPy:

1. calcule o valor do reajuste de cada funcionário;
2. crie um array com os novos salários;
3. calcule o custo total da folha antes do reajuste;
4. calcule o custo total da folha depois do reajuste;
5. descubra quanto a empresa gastará a mais por mês.

**Restrição:** não utilize `for`.

````python
salarios = np.array([1850, 2300, 2750, 3200, 4100, 5250, 6800, 8400])

# Desenvolva sua solução abaixo

reajuste_salario = salarios * 1.057
print("1. Reajuste de salario:", reajuste_salario)
#Custo total antes
print("Custo total da folha antes do reajuste", np.sum(salarios))
#custo total depois do dissidío
print("Custo total da folha depois do reajuste", np.sum(reajuste_salario))
#gasto a mais por mês com o dissidío
print("Gasto da empresa a mais no Mês", np.sum(reajuste_salario - salarios))
````
