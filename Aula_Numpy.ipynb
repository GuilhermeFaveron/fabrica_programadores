# Filtro Booleano (true or false)

```python
array_numeros = np.array ([52, 45, 68, 93, 40, 25, 27])

filtro_valores = array_numeros > 50

array_numeros[filtro_valores]

#filtrando pares
filtro_valores = array_numeros % 2 == 0
print(array_numeros[filtro_valores])
```

```python
#Uma empresa de engenharia possuí as dimensões das peças que irão fabricar diferentes reservatórios, o engenheiro responsável quer calcular o volume dessas peças de formato cúbico.**

#base [2, 1.5, 4, 7, 8]

#largura [3, 5, 2, 3, 3]

#altura [1, 6, 7, 9, 5]

#Calcule o volume de cada peça montada, volume de cubos (multiplicação de todos os lados)

array_base = np.array([2, 1.5, 4, 7, 8])
array_largura = np.array([3, 5, 2, 3, 3])
array_altura = np.array([1, 6, 7, 9, 5])

print(f"O vulume das peças é: {array_base * array_largura * (array_altura)}")

```
